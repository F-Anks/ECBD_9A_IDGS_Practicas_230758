Campo   Descripción
**#**:  El ID o número de identificación del Pokémon. \
**Name**:   El nombre del Pokémon.\
**Type 1**: Se refiere a las propiedades elementales (tipos) asociadas tanto al Pokémon como a sus movimientos.\
**Type 2**: Los Pokémon pueden tener hasta dos tipos (Pokémon de tipo dual); los movimientos solo pueden tener uno.\
**Total**:  Representa la suma de todos los stats, lo que te da una idea de qué tan "pesado" o fuerte es el Pokémon.\
**HP**: (Puntos de Salud) Es qué tanto aguante tiene un Pokémon antes de que lo dejen fuera de combate (se desmaye).\
**Attack**: La fuerza de los ataques físicos del Pokémon.\
**Defense**:    Qué tanto aguanta el Pokémon contra los ataques físicos del rival.\
**Sp. Atk**:    Ataque Especial; el poder de los movimientos especiales del Pokémon.\
**Sp. Def**:    Defensa Especial; qué tan resistente es el Pokémon ante los ataques especiales.\
**Speed**:  Velocidad; el Pokémon que tenga más velocidad es el que ataca primero.\
**Generation**: Se refiere a la generación de la serie de juegos a la que pertenece.\
**Legendary**:  Indica si el Pokémon es Legendario o no (son bichos muy raros y poderosos). Es un valor booleano (sí/no).

### EJERCICIO 6

**Evolve** es una estadística que representa el **si un pokemon puede evolucionar**, mide la capacidad inherente del Pokémon para evolucionar, adaptarse y transformarse en formas más poderosas.
- **Rango:** 10 - 150
- **Cálculo (generado por IA):**
  - Se toma el 15% del Total de estadísticas base como punto de partida
  - Los Pokémon **Legendarios** reciben un bonus de +40 puntos
  - Se añade un factor de variabilidad aleatoria entre -15 y +25
  - El valor final se limita al rango [10, 150]
- **Interpretación:**
  - Valores bajos (10-40): Pokémon con poco potencial evolutivo
  - Valores medios (41-80): Potencial evolutivo moderado
  - Valores altos (81-120): Alto potencial evolutivo
  - Valores muy altos (121-150): Potencial evolutivo excepcional (típico de legendarios)

```python
# Verificar la metadata actualizada
print('Columnas del dataset:', list(pokemons.columns))
print('\nTotal de registros:', len(pokemons))
print('\nTotal de tipos únicos (Type 1):', len(pokemons['Type 1'].unique()))
print('\nTipos disponibles:')
print(pokemons['Type 1'].value_counts())
print('\nEstadísticas de la nueva columna Evolve:')
print(pokemons['Evolve'].describe())
```

```python
# Recalcular los datos y generar los gráficos con los nuevos parámetros
# Ahora incluimos la estadística 'Evolve' y el tipo 'Artificial'

# Actualizar la lista de tipos (ahora incluye 'Artificial')
list_types = pokemons['Type 1'].unique().tolist()
list_types.sort()

# Actualizar las estadísticas a graficar (ahora incluye 'Evolve')
stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed', 'Evolve']

# Paletas de colores (una por estadística, ahora necesitamos 7)
palette = ['Reds', 'Blues', 'Greens', 'Oranges', 'Purples', 'YlOrBr', 'cool']

# Crear la figura con subplots (ahora 7 estadísticas: filas=4, cols=2)
fig, axes = plt.subplots(4, 2, figsize=(25, 40))
fig.suptitle('Total de estadísticas por Tipo de Pokémon (Dataset Modificado)', fontsize=20, fontweight='bold')

k = 0
m = 0
for i in stats:
    row = k // 2
    col = k % 2
    plt.subplot(4, 2, k + 1)
    k = k + 1
    
    # Agrupamos por la columna correcta y sumamos solo valores numéricos
    datos_grafico = pokemons.groupby('Type 1').sum(numeric_only=True)[i]
    
    # Dibujamos el gráfico de barras horizontales
    sns.barplot(x=datos_grafico, y=list_types, palette=palette[m], hue=list_types, legend=False)
    m = m + 1
    
    # Añadimos el título correspondiente a cada estadística
    plt.title(f'Total of {i}')

# Ocultar el subplot vacío (posición 8)
axes[3][1].set_visible(False)

plt.tight_layout()
plt.subplots_adjust(top=0.95)
plt.show()
```

```python
plt.figure(figsize=(13, 30))

for k, stat in enumerate(stats, start=1):

    plt.subplot(len(stats), 1, k)

    sns.stripplot(
        data=pokemons,
        x='Type 1',
        y=stat,
        hue='Type 1',
        palette='Dark2',
        legend=False,
        jitter=True
    )

    plt.title(f'Total of {stat} for each type')
    plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
