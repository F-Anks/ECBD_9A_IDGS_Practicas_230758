## 2.3 Estados con la Mayor Proporción de Donantes Respecto a su Población

Calculamos la proporción de donantes por cada 100,000 habitantes para identificar qué estados tienen una mayor participación relativa en DonorsChoose.

```python
state_counts = donors_df['Donor State'].value_counts()
proportion_data = []

for state, count in state_counts.items():
    if state in us_state_pop:
        prop = (count / us_state_pop[state]) * 100000
        proportion_data.append({
            'Estado': state,
            'Donantes_por_100K': round(prop, 4),
            'Conteo': count,
            'Población': f"{us_state_pop[state]:,}"
        })

prop_df = pd.DataFrame(proportion_data).sort_values('Donantes_por_100K', ascending=False)

fig = go.Figure(go.Bar(
    x=prop_df['Donantes_por_100K'],
    y=prop_df['Estado'],
    orientation='h',
    marker=dict(
        color=prop_df['Donantes_por_100K'],
        colorscale='RdYlGn',
        showscale=True,
        colorbar=dict(title="Por 100K hab.")
    ),
    text=prop_df['Donantes_por_100K'].round(4),
    textposition='outside'
))
fig.update_layout(
    title="Estados con Mayor Proporción de Donantes por 100,000 Habitantes",
    xaxis_title="Donantes por 100,000 habitantes",
    yaxis=dict(autorange="reversed"),
    height=max(450, len(prop_df) * 28),
    margin=dict(l=160, r=80)
)
fig.show()

print("\n📊 Tabla de proporción de donantes por estado:")
display(prop_df.head(10))
```

## 2.4 Visualización de la Ubicación de los Donantes

Utilizamos el dataset de códigos postales para obtener coordenadas geográficas (latitud/longitud) y mapear la ubicación de los donantes en el territorio de EE.UU.

```python
# Preparar datos geográficos
zip_codes_df['zip_code'] = zip_codes_df['zip_code'].astype(str).str.zfill(5)
zip_codes_df['zip_prefix'] = zip_codes_df['zip_code'].str[:3]

# Promediar coordenadas por prefijo de código postal
zip_geo = zip_codes_df.groupby('zip_prefix').agg({
    'latitude': 'mean',
    'longitude': 'mean'
}).reset_index()

# Preparar zip de donantes como cadena de 3 dígitos
donors_df['Donor Zip Str'] = donors_df['Donor Zip'].astype(str).str.zfill(3)

# Merge para obtener coordenadas geográficas
donors_geo = donors_df.merge(
    zip_geo, left_on='Donor Zip Str', right_on='zip_prefix', how='left'
)
donors_geo_clean = donors_geo.dropna(subset=['latitude', 'longitude'])

print(f"✅ Donantes con coordenadas geográficas: {donors_geo_clean.shape[0]} de {donors_df.shape[0]}")
print(f"   ({donors_geo_clean.shape[0]/donors_df.shape[0]*100:.1f}% de cobertura geográfica)")

# Mapa de dispersión general de donantes
fig = go.Figure(data=go.Scattergeo(
    lon=donors_geo_clean['longitude'],
    lat=donors_geo_clean['latitude'],
    mode='markers',
    marker=dict(
        size=8,
        color='#E53935',
        opacity=0.65,
        line=dict(width=0.5, color='white')
    ),
    text=donors_geo_clean['Donor City'] + ', ' + donors_geo_clean['Donor State'],
    hoverinfo='text'
))
fig.update_layout(
    title='🗺️ Ubicación Geográfica de los Donantes en EE.UU.',
    geo=dict(
        scope='usa',
        showland=True,
        landcolor='rgb(243, 243, 243)',
        countrycolor='rgb(204, 204, 204)',
        showlakes=True,
        lakecolor='rgb(255, 255, 255)'
    ),
    height=550
)
fig.show()
```

### 2.4.4 Ubicación de los Donantes en Texas

Visualizamos la ubicación geográfica de los donantes en los 4 estados con mayor presencia: California, Florida, Nueva York y Texas.

```python
# Función para graficar donantes por estado
def plot_state_donors(state_name, color='#1565C0'):
    # Genera un mapa scatter con la ubicación de donantes en un estado específico
    state_data = donors_geo_clean[donors_geo_clean['Donor State'] == state_name]

    if len(state_data) == 0:
        print(f"⚠️ No se encontraron donantes con coordenadas en {state_name}")
        return

    fig = go.Figure(data=go.Scattergeo(
        lon=state_data['longitude'],
        lat=state_data['latitude'],
        mode='markers',
        marker=dict(
            size=12, color=color, opacity=0.75,
            line=dict(width=1, color='white')
        ),
        text=state_data['Donor City'],
        hoverinfo='text'
    ))
    fig.update_layout(
        title=f'🗺️ Donantes en {state_name} ({len(state_data)} donantes)',
        geo=dict(
            scope='usa',
            showland=True,
            landcolor='rgb(243, 243, 243)',
            countrycolor='rgb(204, 204, 204)',
            showlakes=True,
            lakecolor='rgb(255, 255, 255)'
        ),
        height=480
    )
    fig.show()
    cities = state_data['Donor City'].value_counts().head(5)
    print(f"  Top ciudades en {state_name}:")
    for city, cnt in cities.items():
        print(f"    📍 {city}: {cnt} donantes")

# 2.4.1 - California
print("=" * 50)
print("2.4.1 DONANTES EN CALIFORNIA")
print("=" * 50)
plot_state_donors('California', '#1565C0')

# 2.4.2 - Florida
print(f"\n{'='*50}")
print("2.4.2 DONANTES EN FLORIDA")
print("=" * 50)
plot_state_donors('Florida', '#2E7D32')

# 2.4.3 - Nueva York
print(f"\n{'='*50}")
print("2.4.3 DONANTES EN NUEVA YORK")
print("=" * 50)
plot_state_donors('New York', '#E65100')

# 2.4.4 - Texas
print(f"\n{'='*50}")
print("2.4.4 DONANTES EN TEXAS")
print("=" * 50)
plot_state_donors('Texas', '#AD1457')
```

## 2.5 Ubicación General de los Donantes - Mapa de Calor Georeferenciable

Generamos un **mapa de calor (choropleth)** que muestra la concentración de donantes por estado, permitiendo identificar visualmente las regiones con mayor participación en DonorsChoose.

```python
# Mapa Choropleth: Concentración de donantes por estado
state_counts_df = donors_df['Donor State'].value_counts().reset_index()
state_counts_df.columns = ['Estado', 'Conteo']
state_counts_df['Código'] = state_counts_df['Estado'].map(state_abbrev)
state_counts_df = state_counts_df.dropna(subset=['Código'])

fig = go.Figure(data=go.Choropleth(
    locations=state_counts_df['Código'],
    z=state_counts_df['Conteo'],
    locationmode='USA-states',
    colorscale='YlOrRd',
    autocolorscale=False,
    colorbar_title="Nº Donantes",
    text=state_counts_df.apply(
        lambda x: f"{x['Estado']}<br>Donantes: {x['Conteo']}", axis=1
    ),
    hoverinfo='text',
    marker_line_color='white',
    marker_line_width=1.5
))
fig.update_layout(
    title_text='🌡️ Mapa de Calor Georeferenciable: Distribución de Donantes por Estado',
    geo=dict(
        scope='usa',
        projection=dict(type='albers usa'),
        showlakes=True,
        lakecolor='rgb(255, 255, 255)'
    ),
    height=600
)
fig.show()

# Heatmap estático con seaborn (vista complementaria)
plt.figure(figsize=(14, 6))
state_pivot = donors_df.groupby('Donor State').size().reset_index(name='Donantes')
state_pivot = state_pivot.sort_values('Donantes', ascending=False).head(15)
colors_bar = sns.color_palette("YlOrRd", len(state_pivot))
sns.barplot(data=state_pivot, x='Donantes', y='Donor State', palette=colors_bar)
plt.title('Top 15 Estados por Número de Donantes', fontsize=14, fontweight='bold')
plt.xlabel('Número de Donantes')
plt.ylabel('Estado')
plt.tight_layout()
plt.show()
```

## 2.6 Donantes Maestros y No Maestros en los Diferentes Estados

Comparamos la proporción de donantes que son maestros frente a los que no lo son en cada estado, utilizando gráficas de barras apiladas y mapas de calor.

```python
# Tabla cruzada: Estado vs Es Maestro
teacher_state = pd.crosstab(donors_df['Donor State'], donors_df['Donor Is Teacher'])
teacher_state_sorted = teacher_state.sort_values(
    teacher_state.columns.tolist(), ascending=False
).head(20)

# Gráfica de barras apiladas con Plotly
fig = go.Figure()
colors = {'Yes': '#4CAF50', 'No': '#FF5722'}
for col in teacher_state_sorted.columns:
    fig.add_trace(go.Bar(
        name=f'Es Maestro: {col}',
        x=teacher_state_sorted.index,
        y=teacher_state_sorted[col],
        marker_color=colors.get(col, '#999999')
    ))

fig.update_layout(
    barmode='stack',
    title='Donantes Maestros vs No Maestros por Estado (Top 20)',
    xaxis_title='Estado',
    yaxis_title='Número de Donantes',
    xaxis_tickangle=-45,
    height=550,
    legend_title="Tipo de Donante"
)
fig.show()

# Mapa de calor con seaborn
plt.figure(figsize=(8, max(6, len(teacher_state) * 0.35)))
sns.heatmap(
    teacher_state,
    annot=True, fmt='d', cmap='YlOrRd',
    linewidths=0.5, linecolor='white',
    cbar_kws={'label': 'Número de Donantes'}
)
plt.title('Mapa de Calor: Donantes Maestros vs No Maestros por Estado',
          fontsize=13, fontweight='bold')
plt.xlabel('¿Es Maestro?')
plt.ylabel('Estado')
plt.tight_layout()
plt.show()

# Porcentaje de maestros por estado
teacher_pct = teacher_state.div(teacher_state.sum(axis=1), axis=0) * 100
if 'Yes' in teacher_pct.columns:
    teacher_pct_sorted = teacher_pct.sort_values('Yes', ascending=False)
    print("\n📊 Porcentaje de donantes maestros por estado:")
    for state in teacher_pct_sorted.index[:10]:
        pct = teacher_pct_sorted.loc[state, 'Yes']
        print(f"   {state}: {pct:.1f}% maestros")
```

## 2.3 Estados con la Mayor Proporción de Donantes Respecto a su Población

Calculamos la proporción de donantes por cada 100,000 habitantes para identificar qué estados tienen una mayor participación relativa en DonorsChoose.

```python
state_counts = donors_df['Donor State'].value_counts()
proportion_data = []

for state, count in state_counts.items():
    if state in us_state_pop:
        prop = (count / us_state_pop[state]) * 100000
        proportion_data.append({
            'Estado': state,
            'Donantes_por_100K': round(prop, 4),
            'Conteo': count,
            'Población': f"{us_state_pop[state]:,}"
        })

prop_df = pd.DataFrame(proportion_data).sort_values('Donantes_por_100K', ascending=False)

fig = go.Figure(go.Bar(
    x=prop_df['Donantes_por_100K'],
    y=prop_df['Estado'],
    orientation='h',
    marker=dict(
        color=prop_df['Donantes_por_100K'],
        colorscale='RdYlGn',
        showscale=True,
        colorbar=dict(title="Por 100K hab.")
    ),
    text=prop_df['Donantes_por_100K'].round(4),
    textposition='outside'
))
fig.update_layout(
    title="Estados con Mayor Proporción de Donantes por 100,000 Habitantes",
    xaxis_title="Donantes por 100,000 habitantes",
    yaxis=dict(autorange="reversed"),
    height=max(450, len(prop_df) * 28),
    margin=dict(l=160, r=80)
)
fig.show()

print("\n📊 Tabla de proporción de donantes por estado:")
display(prop_df.head(10))
```

## 2.4 Visualización de la Ubicación de los Donantes

Utilizamos el dataset de códigos postales para obtener coordenadas geográficas (latitud/longitud) y mapear la ubicación de los donantes en el territorio de EE.UU.

```python
# Preparar datos geográficos
zip_codes_df['zip_code'] = zip_codes_df['zip_code'].astype(str).str.zfill(5)
zip_codes_df['zip_prefix'] = zip_codes_df['zip_code'].str[:3]

# Promediar coordenadas por prefijo de código postal
zip_geo = zip_codes_df.groupby('zip_prefix').agg({
    'latitude': 'mean',
    'longitude': 'mean'
}).reset_index()

# Preparar zip de donantes como cadena de 3 dígitos
donors_df['Donor Zip Str'] = donors_df['Donor Zip'].astype(str).str.zfill(3)

# Merge para obtener coordenadas geográficas
donors_geo = donors_df.merge(
    zip_geo, left_on='Donor Zip Str', right_on='zip_prefix', how='left'
)
donors_geo_clean = donors_geo.dropna(subset=['latitude', 'longitude'])

print(f"✅ Donantes con coordenadas geográficas: {donors_geo_clean.shape[0]} de {donors_df.shape[0]}")
print(f"   ({donors_geo_clean.shape[0]/donors_df.shape[0]*100:.1f}% de cobertura geográfica)")

# Mapa de dispersión general de donantes
fig = go.Figure(data=go.Scattergeo(
    lon=donors_geo_clean['longitude'],
    lat=donors_geo_clean['latitude'],
    mode='markers',
    marker=dict(
        size=8,
        color='#E53935',
        opacity=0.65,
        line=dict(width=0.5, color='white')
    ),
    text=donors_geo_clean['Donor City'] + ', ' + donors_geo_clean['Donor State'],
    hoverinfo='text'
))
fig.update_layout(
    title='🗺️ Ubicación Geográfica de los Donantes en EE.UU.',
    geo=dict(
        scope='usa',
        showland=True,
        landcolor='rgb(243, 243, 243)',
        countrycolor='rgb(204, 204, 204)',
        showlakes=True,
        lakecolor='rgb(255, 255, 255)'
    ),
    height=550
)
fig.show()
```

### 2.4.4 Ubicación de los Donantes en Texas

Visualizamos la ubicación geográfica de los donantes en los 4 estados con mayor presencia: California, Florida, Nueva York y Texas.

```python
# Función para graficar donantes por estado
def plot_state_donors(state_name, color='#1565C0'):
    # Genera un mapa scatter con la ubicación de donantes en un estado específico
    state_data = donors_geo_clean[donors_geo_clean['Donor State'] == state_name]

    if len(state_data) == 0:
        print(f"⚠️ No se encontraron donantes con coordenadas en {state_name}")
        return

    fig = go.Figure(data=go.Scattergeo(
        lon=state_data['longitude'],
        lat=state_data['latitude'],
        mode='markers',
        marker=dict(
            size=12, color=color, opacity=0.75,
            line=dict(width=1, color='white')
        ),
        text=state_data['Donor City'],
        hoverinfo='text'
    ))
    fig.update_layout(
        title=f'🗺️ Donantes en {state_name} ({len(state_data)} donantes)',
        geo=dict(
            scope='usa',
            showland=True,
            landcolor='rgb(243, 243, 243)',
            countrycolor='rgb(204, 204, 204)',
            showlakes=True,
            lakecolor='rgb(255, 255, 255)'
        ),
        height=480
    )
    fig.show()
    cities = state_data['Donor City'].value_counts().head(5)
    print(f"  Top ciudades en {state_name}:")
    for city, cnt in cities.items():
        print(f"    📍 {city}: {cnt} donantes")

# 2.4.1 - California
print("=" * 50)
print("2.4.1 DONANTES EN CALIFORNIA")
print("=" * 50)
plot_state_donors('California', '#1565C0')

# 2.4.2 - Florida
print(f"\n{'='*50}")
print("2.4.2 DONANTES EN FLORIDA")
print("=" * 50)
plot_state_donors('Florida', '#2E7D32')

# 2.4.3 - Nueva York
print(f"\n{'='*50}")
print("2.4.3 DONANTES EN NUEVA YORK")
print("=" * 50)
plot_state_donors('New York', '#E65100')

# 2.4.4 - Texas
print(f"\n{'='*50}")
print("2.4.4 DONANTES EN TEXAS")
print("=" * 50)
plot_state_donors('Texas', '#AD1457')
```

## 2.5 Ubicación General de los Donantes - Mapa de Calor Georeferenciable

## 2.6 Donantes Maestros y No Maestros en los Diferentes Estados

Comparamos la proporción de donantes que son maestros frente a los que no lo son en cada estado, utilizando gráficas de barras apiladas y mapas de calor.

```python
# Tabla cruzada: Estado vs Es Maestro
teacher_state = pd.crosstab(donors_df['Donor State'], donors_df['Donor Is Teacher'])
teacher_state_sorted = teacher_state.sort_values(
    teacher_state.columns.tolist(), ascending=False
).head(20)

# Gráfica de barras apiladas con Plotly
fig = go.Figure()
colors = {'Yes': '#4CAF50', 'No': '#FF5722'}
for col in teacher_state_sorted.columns:
    fig.add_trace(go.Bar(
        name=f'Es Maestro: {col}',
        x=teacher_state_sorted.index,
        y=teacher_state_sorted[col],
        marker_color=colors.get(col, '#999999')
    ))

fig.update_layout(
    barmode='stack',
    title='Donantes Maestros vs No Maestros por Estado (Top 20)',
    xaxis_title='Estado',
    yaxis_title='Número de Donantes',
    xaxis_tickangle=-45,
    height=550,
    legend_title="Tipo de Donante"
)
fig.show()

# Mapa de calor con seaborn
plt.figure(figsize=(8, max(6, len(teacher_state) * 0.35)))
sns.heatmap(
    teacher_state,
    annot=True, fmt='d', cmap='YlOrRd',
    linewidths=0.5, linecolor='white',
    cbar_kws={'label': 'Número de Donantes'}
)
plt.title('Mapa de Calor: Donantes Maestros vs No Maestros por Estado',
          fontsize=13, fontweight='bold')
plt.xlabel('¿Es Maestro?')
plt.ylabel('Estado')
plt.tight_layout()
plt.show()

# Porcentaje de maestros por estado
teacher_pct = teacher_state.div(teacher_state.sum(axis=1), axis=0) * 100
if 'Yes' in teacher_pct.columns:
    teacher_pct_sorted = teacher_pct.sort_values('Yes', ascending=False)
    print("\n📊 Porcentaje de donantes maestros por estado:")
    for state in teacher_pct_sorted.index[:10]:
        pct = teacher_pct_sorted.loc[state, 'Yes']
        print(f"   {state}: {pct:.1f}% maestros")
```

## 3.2 Proyectos y sus Donaciones

## 3.4 Montos de Donación por Año, Mes, Día y Hora

Analizamos cómo se distribuyen las donaciones a lo largo del tiempo: por año, mes, día de la semana y hora del día.

```python
# Convertir fecha de donación a datetime
donations_df['Donation Received Date'] = pd.to_datetime(donations_df['Donation Received Date'])
donations_df['Year'] = donations_df['Donation Received Date'].dt.year
donations_df['Month'] = donations_df['Donation Received Date'].dt.month
donations_df['Day'] = donations_df['Donation Received Date'].dt.day_name()
donations_df['Hour'] = donations_df['Donation Received Date'].dt.hour

fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        "Donaciones por Año", "Donaciones por Mes",
        "Donaciones por Día de la Semana", "Donaciones por Hora del Día"
    )
)

# Por Año
year_data = donations_df.groupby('Year')['Donation Amount'].agg(['sum', 'count']).reset_index()
fig.add_trace(go.Bar(
    x=year_data['Year'], y=year_data['sum'],
    marker_color='#1565C0', name='Monto por Año',
    text=year_data['sum'].apply(lambda x: f"${x:,.0f}"),
    textposition='outside'
), row=1, col=1)

# Por Mes
month_data = donations_df.groupby('Month')['Donation Amount'].agg(['sum', 'count']).reset_index()
month_names = {1:'Ene', 2:'Feb', 3:'Mar', 4:'Abr', 5:'May', 6:'Jun',
               7:'Jul', 8:'Ago', 9:'Sep', 10:'Oct', 11:'Nov', 12:'Dic'}
month_data['Mes'] = month_data['Month'].map(month_names)
fig.add_trace(go.Bar(
    x=month_data['Mes'], y=month_data['sum'],
    marker_color='#2E7D32', name='Monto por Mes'
), row=1, col=2)

# Por Día de la Semana
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
day_names_es = {'Monday':'Lunes', 'Tuesday':'Martes', 'Wednesday':'Miércoles',
                'Thursday':'Jueves', 'Friday':'Viernes', 'Saturday':'Sábado', 'Sunday':'Domingo'}
day_data = donations_df.groupby('Day')['Donation Amount'].sum().reindex(day_order).reset_index()
day_data['Dia'] = day_data['Day'].map(day_names_es)
fig.add_trace(go.Bar(
    x=day_data['Dia'], y=day_data['Donation Amount'],
    marker_color='#E65100', name='Monto por Día'
), row=2, col=1)

# Por Hora
hour_data = donations_df.groupby('Hour')['Donation Amount'].sum().reset_index()
fig.add_trace(go.Scatter(
    x=hour_data['Hour'], y=hour_data['Donation Amount'],
    mode='lines+markers', marker_color='#AD1457',
    name='Monto por Hora', fill='tozeroy'
), row=2, col=2)

fig.update_layout(
    title_text="Análisis Temporal de las Donaciones",
    height=700, showlegend=False
)
fig.show()

# Conteo de donaciones por período
print("\n📊 Resumen de donaciones por período temporal:")
print(f"  Años con donaciones: {donations_df['Year'].nunique()}")
print(f"  Meses con donaciones: {donations_df['Month'].nunique()}")
print(f"  Año con más donaciones: {year_data.loc[year_data['count'].idxmax(), 'Year']} ({year_data['count'].max()} donaciones)")
```

## 6.3 Proveedores Más Populares y Costo Promedio Asociado

Analizamos cuáles son los proveedores que aparecen con mayor frecuencia y su costo promedio por artículo.

```python
# Proveedores más populares
vendor_stats = resources_df.groupby('Resource Vendor Name').agg(
    Frecuencia=('Project ID', 'count'),
    Costo_Promedio=('Resource Unit Price', 'mean'),
    Costo_Total=('Total Cost', 'sum'),
    Articulos_Unicos=('Resource Item Name', 'nunique')
).reset_index().sort_values('Frecuencia', ascending=False)

top_vendors = vendor_stats.head(15)

# Gráfica de barras doble: frecuencia y costo promedio
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Proveedores por Frecuencia", "Costo Promedio por Proveedor")
)

fig.add_trace(go.Bar(
    x=top_vendors['Frecuencia'],
    y=top_vendors['Resource Vendor Name'].str[:30],
    orientation='h', marker_color='#1565C0',
    text=top_vendors['Frecuencia'], textposition='outside',
    showlegend=False
), row=1, col=1)

fig.add_trace(go.Bar(
    x=top_vendors['Costo_Promedio'],
    y=top_vendors['Resource Vendor Name'].str[:30],
    orientation='h', marker_color='#E65100',
    text=top_vendors['Costo_Promedio'].apply(lambda x: f"${x:,.2f}"),
    textposition='outside', showlegend=False
), row=1, col=2)

fig.update_layout(
    title_text="Proveedores Más Populares y su Costo Promedio",
    height=550
)
fig.update_yaxes(autorange="reversed")
fig.show()

# Tabla resumen
print("\nTabla resumen de proveedores:")
display(vendor_stats.head(10).round(2))

# Scatter: frecuencia vs costo promedio
fig2 = go.Figure(go.Scatter(
    x=vendor_stats['Frecuencia'],
    y=vendor_stats['Costo_Promedio'],
    mode='markers+text',
    marker=dict(size=vendor_stats['Costo_Total'] / vendor_stats['Costo_Total'].max() * 40 + 5,
                color=vendor_stats['Costo_Total'], colorscale='Viridis',
                showscale=True, colorbar=dict(title="Costo Total")),
    text=vendor_stats['Resource Vendor Name'].str[:20],
    textposition='top center',
    textfont=dict(size=9)
))
fig2.update_layout(
    title="Relación: Frecuencia vs Costo Promedio por Proveedor",
    xaxis_title="Frecuencia (veces mencionado)",
    yaxis_title="Costo Promedio ($)",
    height=500
)
fig2.show()
```
